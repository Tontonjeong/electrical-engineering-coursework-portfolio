# RF calculation checks

RFDH 공개 학습면을 읽으며 다시 계산한 식을 dependency-free module과 비교한다. 저장된 출력은 측정 결과가 아니다.

```python
import importlib.util
from pathlib import Path
module_path = Path('../calculators/rf_calculators.py').resolve()
spec = importlib.util.spec_from_file_location('rf', module_path)
rf = importlib.util.module_from_spec(spec)
spec.loader.exec_module(rf)
```

```python
rf.reflection(100 + 0j, 50), rf.quarter_wave(50, 100, 3.5e9, 6)
```

```python
rf.cascade_noise_figure([(15, 2), (10, 8)]), rf.unequal_wilkinson(50, 1)
```
